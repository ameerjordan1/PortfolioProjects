```python
# First let's import the packages we will use in this project
# You can do this all now or as you need them

import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import matplotlib
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8)

pd.options.mode.chained_assignment = None



# Now we need to read in the data
df = pd.read_csv('/Users/ameerjordan/Desktop/Data/Data Analyst Portfolio Projects/Project #4 Correlation in Python/movies.csv')
```

```python
# Now let's take a look at the data

df
```

```python
# We need to see if we have any missing data
# Let's loop through the data and see if there's anything missing

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
# Since we have missing data, we're going to drop the rows to avoid any interference

df = df.dropna()
```

```python
# Data types for our columns

print(df.dtypes)
```

```python
# Change the data types for columns

df['budget'] = df['budget'].astype('int64')

df['gross'] = df['gross'].astype('int64')
```

```python
df
```

```python
# Create correct year column

df['yearcorrect'] = df['released'].str.extract(r'(\d{4})')

df
```

```python
df.sort_values(by = ['gross'], inplace = False, ascending = False)
```

```python
# Drop any duplicates

df ['company'].drop_duplicates().sort_values(ascending = False)
```
